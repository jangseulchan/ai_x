# 2절. 함수의 실행결과를 반환하는 return

````python
def fibonacci_print(n):
    '''
    매개변수로 들어온 n값 미만의 피보나치 수열을 출력
    ex. n=10이면 0 1 1 2 3 5 8을 출력
    '''
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
    print() # 개행
    
def fibonacci_return(n):
    '''
    매개변수로 들어온 n값 미만의 피보나치 수열을 리스트로 return
    ex. n=10이면 [0 1 1 2 3 5 8]을 return
    '''
    result = []
    a, b = 0, 1
    while a < n:
        result.append(a)
        a, b = b, a+b
    return result
    print() # 개행
````

````python
x = fibonacci_print(100)
print(x)
````

````python
x = fibonacci_return(100)
print(x)
print(type(x))
````

## 3.3 튜플 매개변수를 이용한 가변인수 설정

````python
# 파이썬은 오버로딩(함수의 중복)을 지원하지 않습니다
def add(a, b):
    return a+b
def add(a, b, c):
    return a+b+c
````

````python
add(1, 2, 3, 4)
````

````python
# *args : 매개변수들을 하나의 튜플 변수 args로 전달받음
def add(*args):
    '전달받은 매개변수들의 합을 return'
    # print('args =', args)
    result_tot = 0 # 누적변수
    for data in args:
        result_tot += data
    return result_tot
````

````python
print(add(1))
print(add(1, 2))
print(add(1, 2, 3))
````

## 3.6 인수의 언패킹
### 튜플인수 언패킹
- 튜플 언패킹 : 함수 호출시 매개변수앞에 * 붙임
- 튜플 패킹 : 함수 정의시 매개변수 앞에 * 붙임(튜플매개변수)

````python
def add(*args):
    tot = 0
    for num in args:
        tot += num
    return tot
````

### (2) filter
- map : [1,2,3,4,5] => [2,4,6,8,10] (벡터의 모든 데이터를 가공)
- filter : [1,2,3,4,5] => [2,4] (원하는 정보를 추출)
* cf. 리스트 컴프리헨션

````python
def filter_template(func, L=[]):
    result = []
    for l in L:
        if func(l):
            result.append(l)
    return result
````

````python
def evenChk(value):
    return value%2==0
evenChk(10), evenChk(-9)
````

````python
list_data = [1, 2, 3, 4, 5]
filter_template(lambda value : value%2==0 , list_data)
````

````python
list(filter(lambda value : value%2==0 , list_data))
````

## 4.3 return문에 람다식 이용

````python
# 함수를 return
def make_box(shape): # shape : [l]=>len가 l인 list를 생성하는 함수 return,
                    #          [r,c] => r행 c열인 2차원 list를 생성하는 함수 return
    def box(init_value):
        if len(shape)==1:
            return [init_value]*shape[0]
        elif len(shape)==2:
            return [ [init_value]*shape[1] ] * shape[0]
    return box
````

````python
def make_box(shape):
    if len(shape)==1:
        return lambda init_value : [init_value]*shape[0]
    elif len(shape)==2:
        return lambda init_value : [ [init_value]*shape[1] ] * shape[0]
````
