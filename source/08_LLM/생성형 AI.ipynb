```python
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
llm = ChatOllama(model="llama3.2:1b", 
                 temperature=0)
```

```python
# 나라 -> 음식
country_food_prompt = PromptTemplate(
    template="Tell me one of the most famous foods in {country}.Return the name of the food only",
    input_variables=["country"]
)
output_parser = StrOutputParser()
output_parser.invoke(llm.invoke(country_food_prompt.invoke({"country":"Korea"})))
```

```python
country_food_chain = country_food_prompt | llm | output_parser
country_food_chain.invoke({"country":"Korea"})
```

```python
# 음식 -> 레시피
food_recipe_prompt = PromptTemplate(
    template="Tell me the recipe for {food}.",
    input_variables=["food"]
)
output_parser = StrOutputParser()
output_parser.invoke(llm.invoke(food_recipe_prompt.invoke({"food":"Bibimbap."})))
```

```python
food_recipe_chain = food_recipe_prompt | llm | output_parser
food_recipe_chain.invoke({"food":"Bibimbap."})
```

```python
# 나라 -> (음식명 ->) 레시피
final_chain = country_food_chain | food_recipe_chain
final_chain.invoke({"country":"Korea"})
```
